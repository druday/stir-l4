## Facts for the case for a review (session S2, analysis; reads saved files only)
Rule step3-case-1: among the reviews that may bear on the focus (rule step3-rule-2), those that also name fat in another sense (for example liver fat), and the nearest: title names weight, overweight or obesity, and no fat in another sense. Output: data/derived/step3_case_nearest.csv.

```python
import csv, os, re
import stir_l4_local as L
# Step 3, session S2: facts for the case for a review. Reads data/derived/step3_review_selection_v2.csv (rule
# step3-rule-2, L000155) and the saved PubMed records L000084 only. Rule step3-case-1: among the reviews that may bear
# on the focus, 'also organ fat' = names fat in another sense (for example liver fat); 'nearest' = title names weight,
# overweight or obesity, and no fat in another sense. For the nearest, title and abstract are checked for muscle or
# lean mass, body water and an Asian population with rule step3-rule-2's own patterns.
with open(os.path.join(k._ROOT, 'data', 'derived', 'step3_review_selection_v2.csv'), encoding='utf-8') as f:
    may = [r for r in csv.DictReader(f) if r['group'] == 'may bear on the focus']
WEIGHT_TITLE = re.compile(r'\b(weight|overweight|obes\w*)\b', re.I)
organ = [r for r in may if r['other_sense_terms']]
nearest = [r for r in may if not r['other_sense_terms'] and WEIGHT_TITLE.search(r['title'])]
recs = {r['pmid']: r for r in L.pubmed_records('L000084')}
case_rows = []
for r in nearest:
    t = (recs[r['pmid']]['title'] or '') + ' ' + (recs[r['pmid']]['abstract'] or '')
    case_rows.append({'pmid': r['pmid'], 'doi': r['doi'], 'title': r['title'], 'year': r['year'],
                      'search_end': r['search_end_stated'] or 'not stated in the abstract (abstract only: unknown)',
                      'search_end_passage': r['search_end_passage'],
                      'names_muscle_or_lean': bool(L.MUSCLE.search(t)), 'names_body_water': bool(L.WATER_OWN.search(t)),
                      'names_asian': bool(L.ASIAN.search(t)),
                      'names_without_diabetes': bool(L.NO_DIABETES.search(t) or L.NO_DIABETES_2.search(t))})
out_path = os.path.join(k._ROOT, 'data', 'derived', 'step3_case_nearest.csv')
with open(out_path, 'w', newline='', encoding='utf-8') as f:
    w = csv.DictWriter(f, fieldnames=list(case_rows[0]))
    w.writeheader()
    w.writerows(case_rows)
case = {'may_bear': len(may), 'also_organ_fat': len(organ), 'nearest': [r['pmid'] for r in case_rows],
        'nearest_muscle_or_lean': sum(r['names_muscle_or_lean'] for r in case_rows),
        'nearest_body_water': sum(r['names_body_water'] for r in case_rows),
        'nearest_asian': sum(r['names_asian'] for r in case_rows),
        'nearest_without_diabetes': sum(r['names_without_diabetes'] for r in case_rows),
        'nearest_search_end': {r['pmid']: r['search_end'] for r in case_rows}}
print(case)
```

## Facts for the case for a review, broadened focus (session S2, analysis; reads saved files only)
Rule step3-case-2: among the reviews that may bear on the broadened focus (rule step3-rule-3, L000259), the 'weight reviews' are those whose title names overweight or obesity; for each, the focus parts named, whether it also covers people with diabetes and its search end as stated in the abstract. Also counts reviews naming heart failure or kidney disease in the title, the review with the latest stated search end, and, from rule step3-sound-2 (L000262), the trial and any-kind review counts used in the case. Output: data/derived/step3_case_v2.csv.

```python
import csv, json, os
# Step 3, session S2: facts for the case for a review, broadened focus (all adults without diabetes, whatever the drug
# is given for). Reads saved derived files only: data/derived/step3_review_selection_v3.csv (rule step3-rule-3,
# L000259), data/derived/step3_soundness_v2.json (rule step3-sound-2, L000262) and data/derived/step5_records.csv
# (rule step5-rule-1, L000202). Rule step3-case-2: 'weight reviews' = reviews that may bear on the focus whose title
# names overweight or obesity (the conditions_in_title column of rule step3-rule-3); for each, the focus parts it names
# (title and abstract, rule step3-rule-2 patterns, as columns of the same file), whether it also covers people with
# diabetes (mixed_diabetes) and its search end as stated in the abstract.
with open(os.path.join(k._ROOT, 'data', 'derived', 'step3_review_selection_v3.csv'), encoding='utf-8') as f:
    may = [r for r in csv.DictReader(f) if r['group'] == 'may bear on the focus']
T = lambda v: v == 'True'
weight = [r for r in may if 'overweight or obesity' in r['conditions_in_title']]
hf_ckd = [r for r in may if any(c in r['conditions_in_title'] for c in ('heart failure', 'kidney disease'))]
latest = max((r for r in may if r['search_end_sort']), key=lambda r: r['search_end_sort'])
sound = json.load(open(os.path.join(k._ROOT, 'data', 'derived', 'step3_soundness_v2.json'), encoding='utf-8'))
rows = [{'pmid': r['pmid'], 'doi': r['doi'], 'year': r['year'], 'title': r['title'],
         'names_fat': T(r['names_fat']), 'names_muscle_or_lean': T(r['names_muscle_or_lean']),
         'names_water': T(r['names_water']), 'also_with_diabetes': T(r['mixed_diabetes']),
         'search_end': r['search_end_stated'] or 'not stated in the abstract (abstract only: unknown)'} for r in weight]
out_path = os.path.join(k._ROOT, 'data', 'derived', 'step3_case_v2.csv')
with open(out_path, 'w', newline='', encoding='utf-8') as f:
    w = csv.DictWriter(f, fieldnames=list(rows[0]))
    w.writeheader()
    w.writerows(rows)
case2 = {
    'may_bear': len(may),
    'weight_reviews': len(weight),
    'weight_years': sorted((r['year'] for r in rows), reverse=True),
    'weight_names_fat': sum(r['names_fat'] for r in rows),
    'weight_names_muscle_or_lean': sum(r['names_muscle_or_lean'] for r in rows),
    'weight_names_water': sum(r['names_water'] for r in rows),
    'weight_also_with_diabetes': [r['pmid'] or r['doi'] for r in rows if r['also_with_diabetes']],
    'weight_search_end': {(r['pmid'] or r['doi']): r['search_end'] for r in rows},
    'weight_search_end_stated': sum(1 for r in rows if not r['search_end'].startswith('not stated')),
    'hf_or_ckd_in_title': len(hf_ckd),
    'latest_search_end': {'id': latest['pmid'] or latest['doi'], 'end': latest['search_end_stated'],
                          'title': latest['title'], 'names_fat': T(latest['names_fat']),
                          'names_muscle_or_lean': T(latest['names_muscle_or_lean']),
                          'names_water': T(latest['names_water']), 'focus_terms': latest['focus_terms']},
    'trials_rct_tagged': sound['trials']['n'],
    'trials_names_all_three': sum(1 for x in sound['trial_rows'] if x['fat'] and x['muscle_or_lean'] and x['water']),
    'reviews_any_kind': sound['reviews_any_kind']['reviews'],
    'reviews_any_kind_muscle': sound['reviews_any_kind']['names_muscle_or_lean'],
    'reviews_any_kind_muscle_systematic': sound['reviews_any_kind']['muscle_systematic_tag'],
}
print(json.dumps(case2, indent=1))
```
