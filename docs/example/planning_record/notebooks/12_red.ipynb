```python
# A placebo-controlled pilot study in obese adults without diabetes that measured body water and fat (found by the
# mechanism search): is it tagged as a randomized controlled trial, and is it among the Feasibility trial records?
tg = L.record_tags(['29235586'], 'L000196', ['L000193', 'L000195'],
    why='the trial records counted so far are those PubMed tags Randomized Controlled Trial')
print(tg['log_id'], tg['rows'])
```

```python
# Passages quoted with the draft atlas (theory and earlier evidence), checked against the saved abstracts
# (declared normalization: hyphenated line breaks joined, whitespace collapsed).
RED_PASSAGES = [
    ('29235586', 'Glycosuria provides an energy and osmotic drain that could alter body composition',
     'theory: sugar lost in urine may change fat and water'),
    ('29235586', 'There was a reduction in TBW measured by D2O dilution in the RE group that was significantly greater than placebo',
     'earlier evidence: body water fell, obese adults without diabetes'),
    ('42730799', 'compensatory hyperphagia limits the degree of weight loss',
     'theory: eating more offsets part of the loss'),
    ('42730799', 'transiently promotes osmotic diuresis and natriuresis',
     'theory: water and salt loss, early'),
    ('40707349', 'Controversial data are reported for muscle-mass, which decreased in 15 and remained unchanged in 6 studies',
     'earlier review: muscle mass'),
    ('40707349', 'with diabetes-mellitus (68 %) and Japanese (53 %)',
     'earlier review: who the muscle studies enrolled'),
    ('40707349', 'There is an urgent need for research, especially in nondiabetic, non-Japanese and HF-patients',
     'earlier review: where research is needed'),
    ('33357505', 'Fluid volume changes might be an important mechanism underlying the beneficial clinical effects of SGLT2 inhibitors',
     'theory: water in heart failure'),
    ('35478433', 'dapagliflozin decreased BP but did not increase 24-hour sodium and volume excretion',
     'earlier evidence: water and salt in kidney disease without diabetes'),
]
chk = L.check_passages(RED_PASSAGES, ['L000129', 'L000196'], out='data/derived/step7_red_passages.csv')
lid_p = k.log('computation', element='RED', derived_path=chk['path'], refers_to=['L000129', 'L000196'],
    text=f"Step 7, session S3 (the atlas): {len(chk['rows'])} passages on theory and earlier evidence, checked "
         f"against the saved abstracts ({'all found' if chk['all_found'] else 'not all found'}); "
         "declared normalization: hyphenated line breaks joined, whitespace collapsed",
    parameters={'rule': 'passage-check (L._norm_ws)', 'all_found': chk['all_found'],
                'pmids': [r['pmid'] for r in chk['rows']]})
print(chk['all_found'], [(r['pmid'], r['found_in_abstract'], r['source_log_id']) for r in chk['rows']], lid_p)
```
