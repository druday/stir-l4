```python
import stir_l4_local as L
# Session S3, Step 7: rule step7-rule-3 over the saved PubMed records (L000196) of the Feasibility trials search
# (L000195), records tagged 'Randomized Controlled Trial'. Grain: PubMed records. Patterns are my suggestion,
# logged with the computation.
f7 = L.lat_facts('L000196', 'L000195')
print(f7['log_id'], f7['records'], f7['drugs'], f7['crossover'], f7['diet_exercise_lifestyle'],
      f7['first_author_country'], 'passages found:', f7['passages_found'])
```

```python
# Passages quoted with the draft features (treatment time while a comparison group lasts; treatment both groups get),
# checked against the saved abstracts (declared normalization: hyphenated line breaks joined, whitespace collapsed).
LAT_PASSAGES = [
    ("28345814", "entered an open-label extension from 24 to 52 weeks during which all participants received active treatment",
     "treatment time: the comparison ends before the treatment does"),
    ("27550386", "Participants were instructed to follow a balanced diet and exercise moderately",
     "other treatment both groups get: diet or exercise advice"),
]
chk_l = L.check_passages(LAT_PASSAGES, ['L000196'], out='data/derived/step7_lat_passages.csv')
lid_l = k.log('computation', element='LAT', derived_path=chk_l['path'], refers_to=['L000196', f7['log_id']],
    text=f"Step 7, session S3 (features to count): {len(chk_l['rows'])} passages quoted with the draft features, "
         f"checked against the saved abstracts ({'all found' if chk_l['all_found'] else 'not all found'}); "
         "declared normalization: hyphenated line breaks joined, whitespace collapsed",
    parameters={'rule': 'passage-check (L._norm_ws)', 'all_found': chk_l['all_found'],
                'pmids': [r['pmid'] for r in chk_l['rows']]})
print(chk_l['all_found'], [(r['pmid'], r['found_in_abstract']) for r in chk_l['rows']], lid_l)
```
