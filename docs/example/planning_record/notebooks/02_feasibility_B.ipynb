```python
import re
import stir_l4_local as L
# Ongoing studies (ClinicalTrials.gov page saved in L000198), rule step5-rule-2 (rule 1, L000201, was superseded
# before it was shown: it counted pericoronary adipose tissue as body fat). Drug names listed in the PubMed drug block are read
# from the logged string of search L000117, so the comparison uses the block exactly as run.
BLOCK_NAMES = re.findall(r'(\w+gliflozin)\[tiab\]', next(r for r in k.records() if r['id'] == 'L000117')['parameters']['term'])
ong = L.ongoing_studies('L000198', out='data/derived/step5_ongoing_studies.csv', block_names=BLOCK_NAMES,
                         rule=L.RULE_5_2)
print(ong['log_id'], ong['counts'])
```

```python
import json, os
# Known-study check: the trial in people without type 2 diabetes with a Japanese affiliation (PMID 33879970, saved in
# L000129). Is it in the broader search's saved results, does its saved abstract contain the no-diabetes words, and
# what did the phrase-only check (L000200) return?
ab = L.record_details('L000129')['33879970']['abstract']
known = {'pmid': '33879970', 'in_broader_search': '33879970' in L.query_ids('L000193'),
         'in_trials_search': '33879970' in L.query_ids('L000195'),
         'abstract_has_without_type_2_diabetes': 'without type 2 diabetes' in L._norm_ws(ab).lower(),
         'passage': 'NAFLD patients without type 2 diabetes mellitus were enrolled',
         'passage_found': 'NAFLD patients without type 2 diabetes mellitus were enrolled' in L._norm_ws(ab),
         'phrase_check_count': next(r for r in k.records() if r['id'] == 'L000200')['n_total']}
with open(os.path.join(k._ROOT, 'data', 'derived', 'step5_known_study.json'), 'w', encoding='utf-8') as f:
    json.dump(known, f, indent=1)
kid = k.log('computation', element='FE', derived_path='data/derived/step5_known_study.json',
            refers_to=['L000129', 'L000193', 'L000195', 'L000200'],
            text='Known-study check for PMID 33879970: in the broader search? in the trials search? abstract has the no-diabetes words? phrase check count',
            parameters=known)
print(kid, known)
```

```python
import csv, json, os, re
import stir_l4_local as L
# Who "all adults without diabetes" takes in (S2, after Feasibility was certified as broaden): the condition named in
# the title of each record tagged as a randomized controlled trial in the trials search (L000195; records fetched in
# L000196; tags as computed in L000202). Rule step5-rule-3, the assistant's suggestion (a DRAFT): condition words in the
# title only; "risk of heart failure" does not count as heart failure; a title may name more than one condition;
# abstracts are not read for this. Focus words are those already computed by rule step3-rule-2 (step5_records.csv).
COND_RULE = 'step5-rule-3'
CONDITIONS = [
    ('overweight or obesity', r'\bobes\w*|\boverweight\b'),
    ('fatty liver', r'fatty liver|\bNAFLD\b|\bNASH\b|\bMASLD\b|\bMASH\b|steatohepatitis|steatotic liver'),
    ('heart failure', r'(?<!risk of )\bheart failure\b|\bHF[rpm]?EF\b'),
    ('kidney disease', r'kidney disease|\bCKD\b|renal (?:disease|impairment|failure)|dialysis|kidney transplant\w*'),
    ('polycystic ovary syndrome', r'polycystic ovary|\bPCOS\b'),
    ('prediabetes or insulin resistance', r'pre-?diabet\w*|insulin[- ]resistan\w*'),
]
recs = L.record_details('L000196')
with open(os.path.join(k._ROOT, 'data', 'derived', 'step5_records.csv'), encoding='utf-8') as fh:
    s5 = [r for r in csv.DictReader(fh) if r['fetched'] == 'True']
rct = [r for r in s5 if r['in_trials_search'] == 'True' and r['rct_tag'] == 'True']
cond_rows = []
for r in rct:
    title = recs[r['pmid']]['title']
    named = [nm for nm, rx in CONDITIONS if re.search(rx, title, re.I)]
    cond_rows.append({'pmid': r['pmid'], 'year': r['year'], 'title': title, 'conditions': '; '.join(named) or 'none of these',
                      'focus_terms': r['focus_terms'], 'rule': COND_RULE})
cond_path = 'data/derived/step5_conditions.csv'
with open(os.path.join(k._ROOT, *cond_path.split('/')), 'w', newline='', encoding='utf-8') as fh:
    w = csv.DictWriter(fh, fieldnames=list(cond_rows[0].keys()))
    w.writeheader()
    w.writerows(cond_rows)
cond_counts = {nm: sum(1 for x in cond_rows if nm in x['conditions'].split('; ')) for nm, _ in CONDITIONS}
cond_counts['none of these'] = sum(1 for x in cond_rows if x['conditions'] == 'none of these')
# Passage check: the heart-failure review quoted in this session (PMID 34516288, OpenAlex results saved in L000152)
hf_passage = 'Consequently, there are decreases in plasma volume, body weight, and blood pressure.'
hf_found = any(x.get('pmid') == '34516288' and hf_passage in L._norm_ws(x['abstract']) for x in L.openalex_records('L000152'))
cond_id = k.log('computation', element='FE', derived_path=cond_path, refers_to=['L000195', 'L000196', 'L000202', 'L000152'],
                text=f'Conditions named in the titles of the {len(cond_rows)} records tagged as randomized controlled trials '
                     f'(trials search, people without diabetes, any region), rule {COND_RULE}: '
                     + '; '.join(f'{a} {b}' for a, b in cond_counts.items()),
                parameters={'rule': COND_RULE, 'patterns': dict(CONDITIONS), 'flags': 'IGNORECASE', 'field': 'title only',
                            'set': 'records in the trials search tagged Randomized Controlled Trial (L000202)',
                            'n_records': len(cond_rows), 'counts': cond_counts,
                            'passage_check': {'pmid': '34516288', 'source_log_id': 'L000152', 'passage': hf_passage,
                                              'found': hf_found}})
print(cond_id, len(cond_rows), cond_counts, hf_found)
```
